## 6. KPI Calculations

```python
total_sales = df['Sales'].sum()
print(f"Total Sales: ${total_sales:,.2f}")
```

```python
total_profit = df['Profit'].sum()
print(f"Total Profit: ${total_profit:,.2f}")
```

```python
total_orders = df['Order ID'].nunique()
print(f'Total Orders: {total_orders}')
```

```python
profit_margin = (total_profit / total_sales) * 100
print(f"Profit Margin: {profit_margin:,.2f}%")
```

```python
total_customers = df['Customer ID'].nunique()
print(f'Total Customers: {total_customers}')
```

## 7. Group Analysis

```python
category_analysis = (df.groupby('Category')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False))
category_analysis
```

```python
subcategory_analysis = df.groupby('Sub-Category')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
subcategory_analysis
```

```python
region_analysis = df.groupby('Region')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
region_analysis
```

```python
segment_analysis = df.groupby('Segment')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
segment_analysis
```

```python
correlation = df[['Discount', 'Profit']].corr() 
correlation
```

```python
discount_profit = df.groupby('Discount')[['Profit']].mean()
discount_profit
```

```python
top_products_sales = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
top_products_sales
```

```python
top_products_profit = df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(10)
top_products_profit
```

```python
bottom_products_profit = df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).tail(10)
bottom_products_profit
```

```python
shipmode_analysis = df.groupby('Ship Mode')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
shipmode_analysis
```

```python
state_analysis = df.groupby('State')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
state_analysis
```

## 8. Time Analysis

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month_name()
df['Month_Number'] = df['Order Date'].dt.month
```

```python
yearly_analysis = df.groupby('Year')[['Sales', 'Profit']].sum()
yearly_analysis
```

```python
monthly_analysis = df.groupby(['Year', 'Month_Number', 'Month'])[['Sales', 'Profit']].sum()
monthly_analysis
```

```python
monthly_trend = (
    df.groupby(df['Order Date'].dt.to_period('M'))
      [['Sales', 'Profit']]
      .sum()
)
```

## 9. Visualizations

```python
region_analysis[['Sales', 'Profit']].plot(kind='bar', figsize=(8,5))

plt.title('Sales and Profit by Region')
plt.ylabel('Amount ($)')
plt.xticks(rotation=0)
plt.show()
```

```python
segment_analysis[['Sales', 'Profit']].plot(kind='bar', figsize=(8,5))
plt.title('Sales and Profit by Segment')
plt.ylabel('Amount ($)')
plt.xticks(rotation=0)
plt.show()
```

```python
subcategory_analysis[['Sales', 'Profit']].plot(kind='bar', figsize=(8,5))

plt.title('Sales and Profit by Sub-Category')
plt.ylabel('Amount ($)')
plt.xticks(rotation=45)
plt.show()
```

```python
bottom_products_profit.plot(kind='barh', figsize=(12, 5))

plt.title('Bottom 10 Products by Sales')
plt.xlabel('Amount ($)')
plt.show()
```

```python
discount_profit.plot(kind='bar', figsize=(8,5))

plt.title('Average Profit by Discount')
plt.xlabel('Discount')
plt.ylabel("Average Profit")
plt.axhline(y=0, color='red', linestyle='--')
plt.show()
```

## 10. Insights 

### 1. Technology is the highest-performing category
Technology generated the highest sales (836,154) and profit (145,455), making it the most valuable category for the business.

### 2. The West region is the most profitable market
The West region generated over 108,418 in profit, outperforming all other regions in overall profitability.

### 3. The business experienced strong growth over time
Sales increased from 484,247 in 2014 to 733,215 in 2017, while profit grew from 49,544 to 93,439 indicating sustained business growth.

### 4. High discounts negatively impact profitability
Analysis revealed a negative correlation (-0.22) between discount and profit.Transactions with discounts above 30% consistently resulted in negative profit, with the worst point at 50% discount averaging -$310 loss per transaction.

### 5. Several products and sub-categories are loss-making
Despite generating substantial sales, Tables recorded a loss of 17,725. Products such as the Cubify CubeX 3D Printer and Lexmark MX611dhe also generated significant losses.

## 11. Recommendations

### 1. Increase investment in Technology products
Focus marketing, inventory, and sales efforts on Technology products to maximize revenue and profit growth.

### 2. Replicate successful regional strategies
Analyze the factors driving success in the West region and apply similar strategies to lower-performing regions.

### 3. Support continued business growth
Prepare inventory, staffing, and marketing resources to accommodate increasing demand and sustain growth.

### 4. Review discount policies
Limit excessive discounting and evaluate promotions based on profitability rather than sales volume alone.

### 5. Address loss-making products
Review pricing, costs, and discounting practices for unprofitable products and sub-categories to improve margins.
